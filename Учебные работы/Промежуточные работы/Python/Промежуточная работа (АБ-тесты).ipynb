```python
import pandas as pd
import numpy as np
import datetime as dt
from scipy.stats import ttest_ind
```

## Задание 3.1
Импортируйте файл с данными в Jupyter Notebook и расшифруйте идентификаторы городов с помощью листа city_dict

```python
dts = pd.read_excel('Финальная первичная аттестация. Вариант 1. Python.xlsx',sheet_name= 'Dataset')
city = pd.read_excel('Финальная первичная аттестация. Вариант 1. Python.xlsx',sheet_name= 'city_dict')
```

```python
new_dts = dts.merge(city, on='id_city',how= 'inner')
new_dts
```

## Задание 3.2
Рассчитайте результаты АВ-теста и с помощью статистического критерия сделайте вывод

```python
s, p = ttest_ind(new_dts[new_dts['id_group'] == 0]["nflag_order"], new_dts[new_dts['id_group'] == 1]["nflag_order"])
print ("stat:",s)
print ("p_value:",p)
```

Гипотеза H0 не подтвердилась, так как средние не равны.


## Задание 3.3
Запустите цикл, который пробежится по всем городам в датасете. Выявите те города, в которых есть статистически значимая разница между группами А и В. Создайте переменную типа "список", которая будет хранить эти города.

```python
ttest_new_list = []
for i in new_dts['name_city'].unique():
  s, p = ttest_ind(new_dts[(new_dts['name_city']==i) & (new_dts['id_group']==0)]['nflag_order'], new_dts[(new_dts['name_city']==i) & (new_dts['id_group']==1)]['nflag_order'])
  if p < 0.05:
    ttest_new_list.append(i)
```

## Задание 3.4
Создайте дополнительную колонку со значением 0 или 1 в зависимости от часа заказа.

1 - если час пик (возьмите, например, 7,8,9,10,17,18,19,20), а 0 - все остальные часы.

Проверьте результат АВ-теста по отдельности на пиковых и не-пиковых заказах и сделайте выводы.

```python
new_dts['hour'] = new_dts['time_view'].dt.hour
new_dts
```

```python
time = [7,8,9,10,17,18,19,20]
new_dts['flag_hour_peak'] = np.where(new_dts['hour'].isin(time) ,1,0)
new_dts
```

```python
for cy in new_dts['name_city'].unique():
  print(cy)
  print()
  #Пиковые
  s, p = ttest_ind(new_dts[(new_dts['name_city']==cy) & (new_dts['flag_hour_peak'] == 1) & (new_dts['id_group']==0)]['nflag_order'],
                     new_dts[(new_dts['name_city']==cy) & (new_dts['flag_hour_peak'] == 1) & (new_dts['id_group']==1)]['nflag_order'])
  print ("Пиковые заказы:")
  print ("stat:", s)
  print ("p_value:", p)
  if p < 0.05:
    print("Гипотеза Н0 не верна. Между группами есть разница. Ошибка выборки минимальна.")
  else:
      print("Гипотеза Н0 побеждает. Между группами нет разницы или мало данных.")
  print()
  
  #Не-пиковые
  s_t, p_t = ttest_ind(new_dts[(new_dts['name_city']==cy) & (new_dts['flag_hour_peak'] == 0) & (new_dts['id_group']==0)]['nflag_order'],
                     new_dts[(new_dts['name_city']==cy) & (new_dts['flag_hour_peak'] == 0) & (new_dts['id_group']==1)]['nflag_order'])
  print ("Не пиковые заказы:")
  print ("stat:", s_t)
  print ("p_value:", p_t)
  if p_t < 0.05:
    print("Гипотеза Н0 не верна. Между группами есть разница. Ошибка выборки минимальна.")
  else:
      print("Гипотеза Н0 побеждает. Между группами нет разницы или мало данных.")
  print()
  print()
  print()
 
```
