**Поля итоговой таблицы должны быть следующими:**

• city — город.<br>
• id_trading_point — уникальный идентификатор торговой точки, к которой прикреплен данный пользователь.<br>

- Если результат (поле diff) положительный и разница на самом деле есть (на основании p_value), то это положительный исход.<br>
- Если результат (поле diff) отрицательный и разница на самом деле есть (на основании p_value), то это отрицательный исход.<br>
- Если разницы на самом деле нет (на основании p_value), то это нейтральный исход.<br>
